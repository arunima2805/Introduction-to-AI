```python
#Problem 1
from collections import deque

# 1. Store the graph using an adjacency list
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': [],
    'F': []
}

# 2. BFS Traversal
def bfs(graph, start):
    visited = set()
    queue = deque([start])
    result = []

    while queue:
        node = queue.popleft()

        if node not in visited:
            visited.add(node)
            result.append(node)

            for neighbour in graph[node]:
                if neighbour not in visited:
                    queue.append(neighbour)

    return result


# 3. DFS Traversal
def dfs(graph, start, visited=None, result=None):
    if visited is None:
        visited = set()
    if result is None:
        result = []

    visited.add(start)
    result.append(start)

    for neighbour in graph[start]:
        if neighbour not in visited:
            dfs(graph, neighbour, visited, result)

    return result


# Perform traversals
bfs_order = bfs(graph, 'A')
dfs_order = dfs(graph, 'A')

# 4. Print traversal orders
print("BFS Traversal:", " -> ".join(bfs_order))
print("DFS Traversal:", " -> ".join(dfs_order))


# Draw BFS tree
print("\nBFS Tree:")
print("        A(1)")
print("       /    \\")
print("    B(2)    C(3)")
print("    /  \\      \\")
print(" D(4)  E(5)   F(6)")


# Draw DFS tree
print("\nDFS Tree:")
print("        A(1)")
print("       /    \\")
print("    B(2)    C(5)")
print("    /  \\      \\")
print(" D(3)  E(4)   F(6)")
```

```python
#Problem 2
from collections import deque

# Capacities of the two jugs
capacity_A = 4
capacity_B = 3

# BFS function
def water_jug_bfs():
    start = (0, 0) # Initial state (jug A, jug B)
    # queue stores tuples of (current_state, list_of_steps_to_reach_current_state)
    # Each step in the list is (action_description, a_after_action, b_after_action)
    queue = deque([(start, [])])
    visited = set([start])

    while queue:
        (a, b), current_path = queue.popleft()

        # Check if we have exactly 2 litres in either jug
        if a == 2 or b == 2:
            return current_path

        # List of possible next states and actions
        next_possibilities = []

        # 1. Fill Jug A
        if a < capacity_A: # Only fill if not already full
            next_possibilities.append(((capacity_A, b), "Fill Jug A"))

        # 2. Fill Jug B
        if b < capacity_B: # Only fill if not already full
            next_possibilities.append(((a, capacity_B), "Fill Jug B"))

        # 3. Empty Jug A
        if a > 0: # Only empty if not already empty
            next_possibilities.append(((0, b), "Empty Jug A"))

        # 4. Empty Jug B
        if b > 0: # Only empty if not already empty
            next_possibilities.append(((a, 0), "Empty Jug B"))

        # 5. Pour A -> B
        if a > 0 and b < capacity_B: # Only pour if A has water and B is not full
            amount_to_pour = min(a, capacity_B - b)
            next_possibilities.append(
                ((a - amount_to_pour, b + amount_to_pour), "Pour A -> B")
            )

        # 6. Pour B -> A
        if b > 0 and a < capacity_A: # Only pour if B has water and A is not full
            amount_to_pour = min(b, capacity_A - a)
            next_possibilities.append(
                ((a + amount_to_pour, b - amount_to_pour), "Pour B -> A")
            )

        # Add unvisited next states to the queue
        for next_state, action_description in next_possibilities:
            if next_state not in visited:
                visited.add(next_state)
                # Append the action and the resulting state to the path
                queue.append((next_state, current_path + [(action_description, next_state[0], next_state[1])]))

    return None # No solution found

# Solve the problem
solution_steps = water_jug_bfs()

# Print solution
print("Minimum sequence of steps:")
print("Initial State: (0, 0)")

if solution_steps:
    for i, step in enumerate(solution_steps):
        action, a_val, b_val = step
        print(f"Step {i+1}: {action} -> ({a_val}, {b_val})")
else:
    print("No solution found.")
```

```python
#Problem 3
# Road network represented using an adjacency list
graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D'],
    'C': ['A', 'D'],
    'D': ['B', 'C', 'E'],
    'E': ['D'],
    'F': ['G'],
    'G': ['F']
}


# 1. Check whether a route exists between two cities
def route_exists(graph, start, destination, visited=None):
    if visited is None:
        visited = set()

    # If we reached the destination
    if start == destination:
        return True

    visited.add(start)

    # Visit all connected cities
    for city in graph[start]:
        if city not in visited:
            if route_exists(graph, city, destination, visited):
                return True

    return False


# 2. Find all connected components
def find_components(graph):
    visited = set()
    components = []

    for city in graph:
        if city not in visited:
            component = []

            def dfs(node):
                visited.add(node)
                component.append(node)

                for neighbour in graph[node]:
                    if neighbour not in visited:
                        dfs(neighbour)

            dfs(city)
            components.append(component)

    return components


# Check routes
start_city = 'A'
destination_city = 'E'

if route_exists(graph, start_city, destination_city):
    print(f"Route exists from {start_city} to {destination_city}")
else:
    print(f"No route exists from {start_city} to {destination_city}")


# Check another route
start_city = 'A'
destination_city = 'F'

if route_exists(graph, start_city, destination_city):
    print(f"Route exists from {start_city} to {destination_city}")
else:
    print(f"No route exists from {start_city} to {destination_city}")


# Find connected components
components = find_components(graph)

print("\nConnected Components:")

for i, component in enumerate(components, 1):
    print(f"Component {i}: {component}")
```
